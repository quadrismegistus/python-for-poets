## Setup

Run these three cells first. The topic model takes a minute or two, so start it before you choose your word.

```python
#@title Install and import the `python4poets` package
try:
    import python4poets as p4p
except ImportError:
    # Not installed yet: install it into this notebook's Python, then import
    %pip install -q git+https://github.com/quadrismegistus/python-for-poets
    import python4poets as p4p

import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
#@title Download and open the Hansard sample
df = p4p.load_hansard()
```

```python
#@title Find 50 topics (as in Workbook 4B; this takes a minute or two)
model = p4p.topic_model(df, n_topics=50)
```

```python
#@title TODO: Your word (in lower case)
WORD = 'REPLACE_ME'
```

## 2. Who uses it?

```python
#@title How often, in how many speeches, by how many MPs? And who uses it most?
counts = p4p.count_word_over_time(df, WORD)
uses = counts[counts['count'] > 0]
print(f'"{WORD}" is used {uses["count"].sum():,} times, in {len(uses):,} speeches, by {uses["speaker"].nunique():,} MPs.')
uses['speaker'].value_counts().head(10)
```

∇ *Is this a word of many MPs, or of a few? Look up anyone at the top of the list you don't know: why might they use it so much?*

## 3. When, and on which side?

```python
#@title Your word over time, by party
p4p.plot_word_over_time(df, WORD, color='party_main')
```

```python
#@title Your word over time, by side of the House
p4p.plot_word_over_time(df, WORD, color='side')
```

∇ *Is there a rise, a fall, or a peak? When? Does it follow a party, or a side? (Labour was in government 1997–2010; the Conservatives 1979–1997 and from 2010, with the Liberal Democrats until 2015.)*

## 4. Party, position, period, or gender?

These tables give how often your word is used, per million words, in each combination of two columns. Holding one column constant lets you see whether the other still matters.

```python
#@title Your word by party and side (per million words)
table = counts.groupby(['party_main', 'side'])[['count', 'total']].sum()
table['per_million'] = table['count'] / table['total'] * 1_000_000
table['per_million'].unstack().round(1)
```

∇ *Read across each row: within one party, does the word belong to government or opposition? Read down each column: within one side, does it belong to a party? (An empty cell means that party was never on that side.)*

```python
#@title TODO: Try two other columns: e.g. 'decade' and 'side', or 'gender' and 'party_main'
table = counts.groupby(['REPLACE_ME', 'REPLACE_ME_ALSO'])[['count', 'total']].sum()
table['per_million'] = table['count'] / table['total'] * 1_000_000
table['per_million'].unstack().round(1)
```

∇ *Which explanation survives? If you compared women and men, remember that women's speeches are mostly recent ones: could your difference be about period instead? (You can check: put `counts[counts['year'] >= 2010]` in place of `counts`.)*

## 5. Read it

```python
#@title Your word in context, in 15 speeches chosen at random
p4p.kwic(df, WORD).sample(frac=1, random_state=1).head(15)
```

∇ *What do MPs mean by it? Does it mean the same thing to everyone who uses it, on both sides, in every decade? Does reading it confirm the story the plots told, or complicate it?*

To read a whole speech, use its row number from the table above, e.g. `print(df.loc[12345, 'txt'])`.

## 6. Its topic

```python
#@title Which topic gives your word the most weight? What else is in it?
topic = model.find_topic(WORD)
model.top_words(topic)
```

```python
#@title The speeches with the most of this topic
model.top_texts(topic)
```

```python
#@title The topic over time, by party
model.plot_over_time(topic, color='party_main')
```

```python
#@title The topic by party and side (average share of a speech, in %)
(model.texts.groupby(['party_main', 'side'])[f'topic_{topic}'].mean() * 100).unstack().round(2)
```

∇ *Is your word typical of its topic, or an odd one out? Does the topic tell the same story as the word, or a different one? If different, which do you trust more, and why?*
